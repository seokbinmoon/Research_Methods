1. Import dataset
- See variable description

```python
# !pip install pyfixest

import pandas as pd
import numpy as np
import pyfixest as pf

df = pd.read_excel("/content/Data_ST.xlsx")
```

2. Create treatment and post dummies (DiD)
- CDStraded: 1 if the firm has traded CDS during the sample period, and 0 otherwise.

- CDStrading: 1 after the introduction of CDS, 0 otherwise.
- “CDSintroyear”: the year of CDS introduction

```python
df["CDStraded"] = np.where(df["CDSintroyear"].notna(), 1, 0)
df["CDStrading"] = np.where(
    df["CDSintroyear"].notna() & (df["FYEAR"] >= df["CDSintroyear"]),
    1, 0
)

reg_df = df
```

3. Run regressions (Table 3, column 1, 2, 5, 6)
- Include variables in the regressions.
- Include time, industry (column 1, 5), and firm (column 2, 6) fixed effects.
- Use robust standard errors.

```python
common = ("DebtMat + {ind} + MB + FixAsset + Profit + Size + "
          "Volatility + AbEarn + TaxCredit + LCF + Rated + InvGrade + CPprogram + "
          "CDStraded + CDStrading")


m1 = pf.feols(f"BLev ~ {common.format(ind='ind_BLev')} | FYEAR + SICH3",
              data=reg_df, vcov={"CRV1": "GVKEY"})

m2 = pf.feols(f"BLev ~ {common.format(ind='ind_BLev')} | FYEAR + GVKEY",
              data=reg_df, vcov={"CRV1": "GVKEY"})

m5 = pf.feols(f"MLev ~ {common.format(ind='ind_MLev')} | FYEAR + SICH3",
              data=reg_df, vcov={"CRV1": "GVKEY"})

m6 = pf.feols(f"MLev ~ {common.format(ind='ind_MLev')} | FYEAR + GVKEY",
              data=reg_df, vcov={"CRV1": "GVKEY"})

pf.etable([m1, m2, m5, m6])
```
